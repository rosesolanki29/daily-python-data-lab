```python
#abstraction:- in abstraction we hide implementation details and show only relevent data to use
#abstraction also used to enforce some rules to child class

#for abstraction we use "abc" module #example-abc.py
#1.abc= this is module to achive abstraction
#2.Abc=abstract base class- it provides common structure to child classes
#3.abstractmethod= this is decorator and it is used to create method that should be redefine in child class

from abc import ABC,abstractmethod

class Bank(ABC):       #abstract class

    @abstractmethod
    def withdraw(self):
        pass


class SbiBank(Bank):
    def __init__(self,balance):
        self.__balance=balance     #private variable
 

    def withdraw(self,amount):
        self.__balance=amount
        print("cureent balance after withdrawl of",amount,"is",self.__balance)


    def displaybalance(self):
        print("balance is ",self.__balance)


class UbiBank(Bank):
    pass       #error here because withdraw methode is not defined


s=SbiBank(20000)
s.displaybalance()
s.withdraw(3000)
u=UbiBank()
```

```python
# Types of variables and Methods in OOps
## Types  of methods
# 1. Instance/Normal Method
# 2. class method- We use @classmethod decorator to create class method and pass 
                  # cls as argument in that method
# 3. static method - We use @staticmethod decorator to access normal fuction outside 
                   # class 
class Bank:
    bank_name="This is State Bank of india"     # class variable
    def __init__(self,balance):
        self.__balance=balance     # Private   # instance variable

    def withdraw(self,amount):
        self.__balance-=amount
        print("Current Balance after withdrawl of ",amount,"is",self.__balance)

    def displayBalance(self):
        print("Balance is ",self.__balance)

    @classmethod
    def displayBank(cls):     # Class Method
        print("Bank name is ",cls.bank_name)

    @staticmethod
    def add(a,b):            # static method
        c=a+b                # static Variable
        print(c)

s=Bank(50000)
s.displayBank()
s.add(10,20)
```
