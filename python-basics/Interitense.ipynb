```python
#Inheritense- It is a property of one class to aquire properties and methods of other class
#properties and method of other class.

#single level Inheritance
# One parent → One child

class Parent:
    def displayParent(self):
        print("this is Parent class")

class Child(Parent):
    def displayChild(self):
        print("this is child class")

c=Child()
c.displayChild()
c.displayParent()
```

```python
#multilevel Inheritence
# Grandparent → Parent → Child

class A:
    def displayA(self):
        print("class A")

class B(A):
    def displayB(self):
        print("class B")

class C(B):
    def displayC(self):
        print("class C")

c=C()
c.displayC()
c.displayB()
c.displayA()

```

```python
#Multiple Inheritence
# One child inherits from multiple parents

class A:
    def displayA(self):
        print("class A")

class B:
    def displayB(self):
        print("class B")

class C(A,B):
    def displayC(self):
        print("class C")

c=C()
c.displayC()
c.displayB()
c.displayA()

    
```

```python
#Heirerichal Inheritence

class A:
    def displayA(self):
        print("class A")

class B(A):
    def displayB(self):
        print("class B")

class C(A):
    def displayC(self):
        print("class C")

b = B()
c = C()
b.displayA()
c.displayA()
```

```python
# Hybrid Inheritance
# Combination of two or more types of inheritance
#mro -Methode Resolution order

class A:
    def displayA(self):
        print("class A")


class B(A):
    def displayB(self):
        print("class B")


class C(A):
    def displayC(self):
        print("class C")


class D(B,C):
    def displayD(self):
        print("class D")


d = D()
d.displayA()
d.displayB()
d.displayC()
d.displayD()
```

```python
class A:
    def display(self):
        print("class A")


class B(A):
    def display(self):
        print("class B")


class C(A):
    def display(self):
        print("class C")


class D(B,C):
    pass


d = D()
d.display()
print(D.mro())
print(B.mro())
```
