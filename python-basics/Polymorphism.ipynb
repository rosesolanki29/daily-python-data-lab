```python
#METHOD OVERRIDING- in python,if same method present in a parent and child calss and we create object of child class
#then the method in child class oveeride parents method

class Parent:
    def display(self):
        print("its parent class")

class Child(Parent):
    def display(self):
        print("its child class")

c=Child()
c.display()  #child class's display method override parent's display methode.
    
```

```python
#operator Overloading
class employee:
    def __init__(self,name,salary):
        self.name=name
        self.__salary=salary

    def display(self):
        print("employee name is",self.name,"and salary is",self.__salary)

    def __add__(self,other):
        return self.__salary+other.__salary

e1=employee("rose",50000)
e1.display()

e2=employee("yash",600000)
e2.display()

print(e1+e2) #e1.__add__(e2) #e1.salary
```
