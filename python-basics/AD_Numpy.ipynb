```python
import numpy as np

#iteration
arr=np.arange(10,15)
print(arr)
for i in arr:
    print(i)
```

```python
#indirect iteration
arr=np.arange(10,15)
print(arr)
for i in range(len(arr)):
    print(i,arr[i])
```

```python
arr=np.arange(10,15)
print(arr)
print(len(arr))
count=0
for i in arr:
    count=count+1
print(count)
```

```python
arr=np.arange(10,19).reshape(3,3)
print(arr)
print()
for i in arr:
    for j in i:
        print(j)
```

```python
#np.nditer(arrayobj)
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for i in np.nditer(arr):
    print(i)
```

```python
#np.ndenumerate(arrayobj)- return index and value
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for index,value in np.ndenumerate(arr):
    print(index,value)
```
