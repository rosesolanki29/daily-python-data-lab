```python
#set-collection of elements
#denoted by {} with csv
#unordered,unindexed,mutable and heterogenous in nature
#store only unique elements
s=set()  #empty set with the help of set constructor
print(s)
print(type(s))
```

```python
#add()
#setobj.add(elements)
s={10,20,30,10,"hello",10.5,10.20,10}
s.add(1000)
print(s)
```

```python
#update()
#setobj.update(iterables)
s={10,20,30,10,"hello",10.5,10.20,10}
s.update({100,200,10,100,200,30})
print(s)
```

```python
s={10,20,30,10,"hello",10.5,10.20,10}
s.update("hello")
print(s)
```

```python
#pop()
#setobj.pop()
s={10,20,30,10,"hello",10.5,10.20,10}
print(s.pop())
print(s)
```

```python
#remove
#setobj.remove(elememts)
s={10,20,30,10,"hello",10.5,10.20,10}
print(s)
s.remove(30)
print(s)
```

```python
s={10,20,30,10,"hello",10.5,10.20,10}
print(s)
s.remove(1000)
print(s)
```

```python
#discard()
#setobj.discard(element)
s={10,20,30,10,"hello",10.5,10.20,10}
print(s)
s.discard(30)
print(s)
```

```python
s={10,20,30,10,"hello",10.5,10.20,10}
print(s)
s.discard(1000)
print(s)
```

```python
s={10,20,30,10,"hello",10.5,10.20,10}
s.clear()
print(s)
```

```python
#set operation
s1={10,50,40,60,80}
s2={20,80,10,70,30}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s1.symmetric_difference(s2))
```

```python
s1={10,50,40,60,80}
s2={20,80,10,70,30}
print(s1|(s2))
print(s1&(s2))
print(s1-(s2))
print(s1^(s2))
```
