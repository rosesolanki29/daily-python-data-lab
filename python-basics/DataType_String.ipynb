```python
#indexing & slicing

#index- position of character in a given string
#1st character of a string -- 0(index)
#last character of any string ---( -1)

#indexing:- is a technique for accesing indiviual character unit for a given string

s="welcome to bhopal"
print(s[0])
print(s[-1])
print(s[-3])
```

```python
#slicing:- technique for accessing multiple character from a given string 
#s[start:stop:step] 
#start - by default (0)
#stop - by default end of the string
# # if stop=n it will run to (n-1)
#step - by default +1


s="welcome to bhopal"
print(s[: :])
print(s[:7:])
print(s[1:11:2])
print(s[-1:-10:-2])
print(s[: :-1])
print(s[-2:-12:-1])
```

```python
#wapp to print number of vowels consonant and the spaces in a given ?
s="welcome to city of lakes"
no_c=0
no_v=0
no_s=0
for i in s:
    if i in "AEIOUaeiou":
        no_v=no_v+1
    elif i==" ":
        no_s=no_s+1
    else:
        no_c=no_c+1
print("number of vowels:",no_v)
print("number of consonants:",no_c)
print("number of space:",no_s)
```

```python
#wapp to check weather a given string is palindrome or not
s=input("enter any string:")
rev=""
for i in s:
    rev=i+rev
if s==rev:
    print(s,"palindrome")
else:
    print(s,"not a palindrome")
```

```python
#index- return position of any character/substring
#strobj.index(character/substring.[start],[stop])
#start=0
#stop- end of the string - n-(n-1)
s="welcome to bhopal"
print(s.index("o"))
print(s.index("o",5))
print(s.index("bhopal"))
print(s.index("o",5,9))
```

```python
#find- return position of given character/substring
#strobj.find(char/substring,[start],[stop])
#start-0,stop-end of the string - n-n-1
s="welcome to bhopal"
print(s.find("o"))
print(s.find("o",5))
print(s.find("bhopal"))
print(s.find("o",5,9))
```

```python
#count-occurence of any given character/substring from a given string
#strobj.count(character/substring,[start],[stop])
s="welcome to bhopal"
print(s.count("o"))
print(s.count("o",5))
print(s.count("bhopal"))
print(s.count("o",5,9))
```
