```python
# direct iteration
arr=np.arange(10,15)
print(arr)
for i in arr:
    print(i)
```

```python
# indirect iteration
arr=np.arange(10,15)
print(arr)
for i in range(len(arr)):
    print(i,arr[i])
```

```python
arr=np.arange(10,15)
print(arr)
print(len(arr))
count=0
for i in arr:
    count=count+1
print(count)
```

```python
arr=np.arange(10,19).reshape(3,3)
print(arr)
print()
for i in arr:
    for j in i:
        print(j)
```

```python
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
print()
for i in arr:
    for j in i:
        for k in j:
            print(k)
```

```python
# np.nditer(arrayobj)
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for i in np.nditer(arr):
    print(i)
```

```python
# np.ndenumerate(arrayobj) - return index and value
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for index,value in np.ndenumerate(arr):
    print(index,value)
```

```python
import numpy as np
l=[10,8,12,20,21,22,21,20,23,23,24,28,29,32,30,32,56,47,67,87,24,25,26,24,25,25,25,26,31,20,20,23,27,28,28,29,26]
age=np.array(l)
print(age)
```

```python
np.mean(age)
```

```python
np.median(age)
```

```python
Q1=np.quantile(age,0.25)
Q1
```

```python
Q2=np.quantile(age,0.50)
Q2
```

```python
Q3=np.quantile(age,0.75)
Q3
```

```python
IQR=Q3-Q1
IQR
```

```python
upper_fence=Q3+(1.5*IQR)
upper_fence
```

```python
lower_fence=Q1-(1.5*IQR)
lower_fence
```

```python
l=[]
for i in age:
    if i<lower_fence or i>upper_fence:
        l.append(i)
print(len(l))
```

```python
# join two array
# concatenate - always return same dimension array with different 
# shape(increase shape)
# dimension and shape both are same
arr1=np.array([10,20,30,40])
arr2=np.array([100,200,300,400])
print(arr1)
print(arr2)
```

```python
arr_new=np.concatenate((arr1,arr2))
print(arr_new)
```

```python
# dimension same but shape are different.
arr1=np.array([10,20,30,40])
arr2=np.array([100,200,300,400,500,600])
print(arr1)
print(arr2)
```

```python
arr_new=np.concatenate((arr1,arr2))
print(arr_new)
```

```python
# different dimension - throw error for concatenate function - arrays  must be same dimension.
arr1=np.array([10,20,30,40])
arr2=np.array([[100,200,300,400]])
print(arr1)
print(arr2)
```

```python
arr_new=np.concatenate((arr1,arr2))
print(arr_new)
```

```python
# 2 dimension - dimension and shape are same
arr1=np.array([[100,200,300],[400,500,600]])
arr2=np.array([[1,2,3],[4,5,6]])
print(arr1)
print(arr2)
```

```python
# by default axis - 0 i.e rows
arr_new=np.concatenate((arr1,arr2))
print(arr_new)
```

```python
arr_new=np.concatenate((arr1,arr2),axis=1)
print(arr_new)
```

```python
# dimension same shape different
arr1=np.array([[100,200],[300,400],[500,600]])
arr2=np.array([[1,2,3],[4,5,6]])
print(arr1)
print(arr2)
arr_new=np.concatenate((arr1,arr2),axis=1)
print(arr_new)
```

```python
arr1=np.array([[100,200,300,400],[300,400,500,600]])
arr2=np.array([[1,2,3],[4,5,6]])
print(arr1)
print(arr2)
arr_new=np.concatenate((arr1,arr2),axis=1)
print(arr_new)
```

```python
# stack
arr1=np.array([10,20,30,40])
arr2=np.array([100,200,300,400])
print(arr1)
print(arr2)
print("----*****----")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("----*****----")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
```

```python
#transpose- arrayobj.T
arr=np.array([[100,200,300,400]])
print(arr)
print(arr.shape)
arr_1=arr.T
print(arr_1)
print(arr_1.shape)
```

```python
arr1=np.array([[10,20,30,40],[50,60,70,80]])
arr2=np.array([[100,200,300,400],[500,600,700,800]])
print(arr1)
print(arr2)
print("----*****----")
arr_new=np.stack((arr1,arr2))
print(arr_new)
print("----*****----")
arr_new=np.stack((arr1,arr2),axis=0)
print(arr_new)
print("----*****----")
arr_new=np.stack((arr1,arr2),axis=1)
print(arr_new)
print("----*****----")
arr_new=np.stack((arr1,arr2),axis=2)
print(arr_new)

```
