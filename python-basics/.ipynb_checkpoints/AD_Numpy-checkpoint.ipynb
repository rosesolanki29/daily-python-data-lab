```python
#direct itration 
arr=np.arange(10,15)
print(arr)
for i in arr:
    print(i)
```

```python
#indirect itration
arr=np.arange(10,15)
print(arr)
for i in range(len(arr)):
    print(i,arr[i])
```

```python
arr=np.arange(10,15)
print(arr)
print(len(arr))
count=0
for i in arr:
    count=count+1
print(count)    
```

```python
arr=np.arange(10,19).reshape(3,3)
print(arr)
print()
for i in arr:
    for j in i:
        print(j)
```

```python
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
print()
for i in arr:
    for j in i:
        for k in j:
            print(k)
```

```python
#np.nditter(array obj)
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for i in np.nditer(arr):
    print(i)
```

```python
#nd.iter(array obj)
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for i in np.nditer(arr):
    print(i)
```

```python
#np.enumarate---- written index and value
arr=np.arange(10,22).reshape(2,2,3)
print(arr)
for index,value in np.ndenumerate(arr):
    print(index,value)
```

```python
l=[20,21,22,21,20,23,23,24,28,29,32,30,32,24,25,26,24,25,25,25,25,26,31,20,20,23,27,28,28,29,26]
age=np.array(l)
print(age)
```

```python
np.mean(age)
```

```python
np.median(age)
```

```python
q1=np.quantile(age,0.25)
q1
```

```python
q2=np.quantile(age,0.50)
q2
```

```python
q3=np.quantile(age,0.75)
q3
```

```python
IQR=q3-q1
IQR
```

```python
upeer_fence=q3+(1.5*IQR)
upeer_fence
```

```python
lower_fence=q1-(1.5*IQR)
lower_fence
```

```python
l=[]
for i in age:
    if i<lower_fence or i>upeer_fence:
        l.append(i)
print(len(l))        
```
