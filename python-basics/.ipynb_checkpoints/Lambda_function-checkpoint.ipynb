```python
# Lambda function is an anonymous function, which means a function
# without a name. It is defined using the 'lambda' keyword instead
# of the traditional 'def' keyword.

# Lambda functions are commonly used for solving short and simple
# problems. They are mainly used with higher-order functions like
# map,filter

# A lambda function can take any number of arguments, but it can
# contain only one expression.

#syntax:- lambda no.of parameter:single exp
```

```python
l=["madam","bhopal","naman","cea","django"]
l1=[]
for i in l:
    rev=""
    for j in i:
        rev=j+rev
    if i==rev:
           l1.append(i)
print(l1)
print([i for i in l if i == i[::-1]])

```

```python
#walf to add 2 number?
a=lambda x,y:x+y
print(a(10,5))
```

```python
#walf to find square of given number?
a=lambda x:x**2
print(a(2))
```

```python
#walf to find largest amoung 2 number?
a=lambda x,y,z:x if x>y and x>z else y if y>x and y>z else z
print(a(10,100,9))
```
