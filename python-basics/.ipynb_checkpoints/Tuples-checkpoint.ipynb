```python
#tuple- collection of elements
#denoted by () with csv
#ordered.indexed,immutable,heterogenous in nature.
t=()
print(t)
print(type(t))
```

```python
#single valued tuple
t=(10,)
print(t)
print(type(t))
```

```python
t=(10,20,30,"hello",10.5)
print(t)
print(type(t))
print(t[0])  #indexed
t[0]=1000   #immutable
print(t)
```

```python
#direct iteration
t=(10,20,30,"hello",10.5)
for i in t:
    print(i)
```

```python
#indirect iteration
t=(10,20,30,"hello",10.5)
for i in range(len(t)):
    print(i,t[i])
```

```python
#functions
#len()- no of elements in a given tuple
t=(10,20,30,"hello",10.5)
print(len(t))
count=0
for i in t:
    count=count+1
print(count)
```

```python
#max()
#min()
t=(10,20,30,10.5)
print(max(t))
print(min(t))
```

```python
t=(10,20,30,10.5)
max_value=t[0]
for i in t:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
t=(10,20,30,10.5)
min_value=t[0]
for i in t:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
#sum
t=(10,20,30,40,10.5)
print(sum(t))
sum=0
for i in t:
    sum=sum+i
print(sum)
```

```python
t1=(10,20,30,40,"hello")
t2=(1000,200,300,"decode",10.5)
t3=t1+t2
print(t3)
```

```python
#sorted(iterable,[reverse=false])
#it always return a list
t=(100,2,34,12,56,27)
print(sorted(t))
print(sorted(t,reverse=True))
```

```python
#zip
t1=(100,200,300,400,500,600)
t2=(1,2,3,4,5)
for i,j in zip(t1,t2):
    print(i,j)
```

```python
#enumerate()
t1=(100,200,300,400,500,600)
for i,j in enumerate(t1):
    print(i,j)
```

```python
#count() -occurence of elements
#tupleobj.count(elements)
t=(10,20,30,10,20,10,"hello",100,10)
print(t.count(10))
print(t.count(20))
print(t.count(1000))
```

```python
#indexed()- position of an elements in a given tuple
#tupleobj.index(elememts,[start],[stop])
t=(10,20,30,40,10,20,30,100,200,10,50)
print(t.index(10))
print(t.index(10,1))
print(t.index(10,5,9))
```

```python
l=[2,3,1,1,2,4,5,5,7,9,5,5,4,7,9,5]
sum=0
count=0
for i in l:
    sum=sum+i
    count=count+1
mean=sum/count
print(mean)
sorted_list=sorted(l)
n=len(l)
if n%2!=0:
    median=sorted_list[n//2]
else:
    median=(sorted_list[n//2]+sorted_list[n//2-1])/2
print(median)

d = {}
for i in l:
    if i in d:
        d[i]+= 1
    else:
        d[i]=1
print(d)
```
