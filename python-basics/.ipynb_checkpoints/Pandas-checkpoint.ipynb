```python
# Pandas :- Pandas is a python package or a library which is used for handling large dataset,exploring
# data , manipulating data ,cleaning data ,transforming data and modeling data.
# their are 2 types of data structure:-
# 1 Series :- single dimension data structure for storing different datatype.
# 2 Data frame :- it is 2 dimensional structure look like a table ,store data in row and column
```

```python
import pandas as pd 
```

```python
l=[10,20,30,40,50]
s=pd.Series(l)
print(s)
print(type(s))
```

```python
s=pd.Series([10,20,30,40,50],index=["a","s","d","e","f"],dtype="float64",name="python")
print(s)
print(s["d"])
```

```python
s=pd.Series({"name":["abc","def","pqr"],"subject":["da","ds","de"],"fees":[100,200,300]})
print(s)
```

```python
s=pd.Series(12)
print(s)
```

```python
s=pd.Series(12,index=[1,2,3,4,5,6,7])
print(s)
```

```python
s1=pd.Series(12,index=[1,2,3,4,5,6,7])
s2=pd.Series(12,index=[1,2,3,4,5])
print(s1+s2)
```

```python
# DataFrame
l=[10,20,30,40,50]
df=pd.DataFrame(l)
df
```

```python
df=pd.DataFrame({"name":["abc","def","pqr"],"subject":["da","ds","de"],"fees":[100,200,300]})
df
```

```python
d={"name":["abc","def","pqr"],"subject":["da","ds","de"],"fees":[100,200,300]}
df=pd.DataFrame(d,columns=["name","fees"],index=["a","s","d"])
df
```

```python
df=pd.DataFrame([[1,2,3,4,5],[10,20,30,40,50]])
df
```

```python
df=pd.DataFrame({"A":pd.Series([1,2,3,4]),"B":pd.Series([10,20,30,40]),1:pd.Series([100,200,300,400])})
df
```

```python
df["C"]=df["A"]+df["B"]
df
```

```python
df["D"]=df["C"]>20
df
```

```python
df.insert(3,"python",df["A"])
df
```

```python
df.insert(3,"python_1",["abc","def","xyz","pqr"])
df
```

```python
# delete
df.pop("python_1")
df
```

```python
del df[1]
df
```
