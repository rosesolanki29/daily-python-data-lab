```python
#oops- object oriented programming system/structure

#it is a programming technique in which we create programs using class and object
#class - It is a blueprint or template
#object - It is a an instance of class

#we create class using keyword class

class House:
    pass

a1=House()
a2=House()
print(a1)
print(a2)
```

```python
#constructor= It used to initialize class object.
#self= It stores object data and handle particular object's data

class Person:
    def __init__(self,name,age):                        #Constructor Method
        self.name=name
        self.age=age
        print("Constructor Called")
p1=Person("Rose",20)
print(p1.age)
print(p1.name)

        
```

```python
class Account:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
account1=Account("rose",120000)
print(account1.name)
print(account1.balance)
account1.balance=-500
print(account1.balance)
```

```python
#Encapsulation
#wraping data and method is single unit (class)and
#restrict direct acces of object's data outside of class.
#access specifier- provide security to object data
    #1.Public variable - self.__name=name
    #2.protected variables- self.__name=name
    #3.private variables -self.__name=name

#getter method and setter method
#getter and setter is used to set or get private variables or used to perform operations on private variables.

class Account:
    def __init__(self,name,balance):
        self.name=name       #public Variables
        self.__balance=balance    #private Variables

    def getbalance(self):
        print("Balance is ",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount

a1=Account("rose",500)
print(a1.name)
a1.getbalance()
a1.setbalance(2000)
a1.getbalance()
```

```python
#dacorators in python?

# A decorator in Python is a function that adds extra functionality to another function without changing its original code.

# Decorators are commonly used for:

# Logging
# Authentication/Authorization
# Timing function execution
# Caching
# Input validation

def decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@decorator
def greet():
    print("Hello!")

greet()
```

```python
#method 2 of getter and setter
#we use @property decorator

class Account:
    def __init__(self,name):
        self.__name=name

    @property
    def name(self):
        print(self.__name)

    @name.setter
    def name(self,newname):
        self.__name=newname

a=Account("rose")
print(a.name)
a.name="roz"
print(a.name)
```
