```python
#dictionary comprehension
#{k:v looping statements condition statements}
l=[1,2,3,4,5]
#{1:1,2:8,3:27,4:64,5:125}
d={}
for i in l:
    d[i]=i**3
print(d)
print({i:i**3 for i in l})
```

```python
l=[1,2,3,4,5]
#{"A":1,"c":9,"E":25}
d={}
for i in l:
    if i%2!=0:
        d[chr(64+i)]=i**2
print(d)
print({chr(64+i):i**2 for i in l if i%2!=0})
```

```python
# Q1. Frequency Mapping with Conditions
# Given a list of words:
# words = ["data", "science", "AI", "python", "data", "code", "python"]

# Write a Python program to create a dictionary where:
# - Keys are words
# - Values are their frequencies
# - Include only those words whose length is greater than 3
# Do not use the count() method.
# - by dictionary comprehension -you can use count method.

l=["data","science","AI","python","data","code","python"]
d={}
for i in l:
    if len(i)>3:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
print(d)

print({i:l.count(i) for i in l if len(i)>3})
```
