```python
# dictionary - collection of elements 
# denoted by {k:v}
# keys must be unique and values can be duplicate
# ordered,unindexed,mutable
d={}
print(d)
print(type(d))
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
print(type(d))
print(d["name"])  # unindexed
d["name"]="aman"  # mutable
print(d)
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
d["Name"]="aman"
print(d)
```

```python
d={}
d["name"]="decode"
print(d)
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
d["name"].append("aman")
print(d)
d["name"].pop()
print(d)
```

```python
# iteration
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
for i in d:
    print(d[i])
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
for i in d:
    for j in d[i]:
        print(j)
```

```python
l=[1,2,3,4,5]
d={}
for i in l:
    if i%2!=0:
         d[chr(64+i)]=i**3
print(d)
    
```

```python
l1=[5,1,6,4,3,2]
l2=[2,6,1,3,4,5]
d={}
for i,j in zip(l1,l2):
    d[i]=i**j
print(d)
```

```python
l=["red","brown","red","green","red","brown","Green"]
d={}
for i in l:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
print(d)
```

```python
#methods
#dictobj.get(key)
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
print(d.get("name"))
```

```python
#keys()
#dictobj.keys()
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
print(d.keys())
for i in d.keys():
    print(i)
```

```python
#values()
#dictobj.values()
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
print(d.values())
```
