```python
# dictionary - collection of elements 
# denoted by {k:v}
# keys must be unique and values can be duplicate
# ordered,unindexed,mutable
d={}
print(d)
print(type(d))
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
print(type(d))
print(d["name"])  # unindexed
d["name"]="aman"  # mutable
print(d)
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
d["Name"]="aman"
print(d)
```

```python
d={}
d["name"]="decode"
print(d)
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
d["name"].append("aman")
print(d)
d["name"].pop()
print(d)
```

```python
# iteration
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
for i in d:
    print(d[i])
```

```python
d={"name":["a","b","c"],"course":["da","ds","de"],"fees":[30000,40000,50000]}
print(d)
for i in d:
    for j in d[i]:
        print(j)
```
