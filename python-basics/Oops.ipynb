```python
#constructor= It used to initialize class object.
#self= It stores object data and handle particular objec's data

class Person:
    def __init__(self,name,age):                        #Constructor Method
        self.name=name
        self.age=age
        print("Constructor Called")
p1=Person("Rose",20)
print(p1.age)
print(p1.name)

        
```

```python
class Account:
    def __init__(self,name,balance):
        self.name=name
        self.balance=balance
account1=Account("rose",120000)
print(account1.name)
print(account1.balance)
account1.balance=-500
print(account1.balance)
```

```python
#Encapsulation
#wraping data and method is single unit (class)and
#restrict direct acces of object's data outside of class.
#access specifier- provide security to object data
    #1.Public variable - self.__name=name
    #2.protected variables- self.__name=name
    #3.private variables -self.__name=name
class Account:
    def __init__(self,name,balance):
        self.name=name       #public Variables
        self.__balance=balance    #private Variables

    def getbalance(self):
        print("Balance is ",self.__balance)

    def setbalance(self,amount):
        self.__balance=amount

a1=Account("rose",500)
print(a1.name)
a1.getbalance()
a1.setbalance(2000)
a1.getbalance()
```
