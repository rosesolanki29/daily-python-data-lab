```python
#wap function that return an arms strong no range b/w start and stop?
def arms(start,stop):
    l=[]
    for i in range(start,stop+1):
            a=len(str(i))
            old=i
            sumdigit=0
            while i!=0:
                dig=i%10
                sumdigit=sumdigit+(dig**a)
                i=i//10
            if sumdigit==old:
                l.append(old)
    return l
```

```python
start=int(input("enter any number: "))
stop=int(input("enter any number: "))
arms(start,stop)
```
