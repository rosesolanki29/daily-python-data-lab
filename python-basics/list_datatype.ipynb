```python
#list-collection of elements
#denoted by [] with csv
#ordered,indexed,mutable and hetrogenous
l=[]  #empty
print(l)
print(type(l))
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l)
print(type(l))
print(l[0])  #indexed
l[1]=20000    #mutable- does support item assignment
print(l)
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l[-1])
print(l[-2])
print(l[-2][1])
print(l[3])
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l[::])
print(l[1:52:])
print(l[:4:])
print(l[::-1])   #step-1 , by default start= -1
```

```python
#iteration
#a) direct iteration
l=[10,20,30,10,50,"hello",10.5]
for i in l:
    print(i)
```

```python
#indirect iteration----with help of range function
l=[10,20,30,"hello",200.5]
for i in range (len(l)):
    print(i,l[i])
```
