```python
#list-collection of elements
#denoted by [] with csv
#ordered,indexed,mutable and hetrogenous
l=[]  #empty
print(l)
print(type(l))
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l)
print(type(l))
print(l[0])  #indexed
l[1]=20000    #mutable- does support item assignment
print(l)
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l[-1])
print(l[-2])
print(l[-2][1])
print(l[3])
```

```python
l=[10,20,30,10,50,"hello",10.5]
print(l[::])
print(l[1:52:])
print(l[:4:])
print(l[::-1])   #step-1 , by default start= -1
```

```python
#iteration
#a) direct iteration
l=[10,20,30,10,50,"hello",10.5]
for i in l:
    print(i)
```

```python
#indirect iteration----with help of range function
l=[10,20,30,"hello",200.5]
for i in range (len(l)):
    print(i,l[i])
```

```python
#functions
#len() - return no of elements in a given list
l=[10,20,100,10,"hello",10.5]
print(len(l))
count=0
for i in l:
    count=count+1
print(count)
```

```python
#max()
l=[10,2,90,10.5]
print(max(l))
print(min(l))
```

```python
l=[10,2,90,10.5]
max_value=l[0]
for i in l:
    if i>max_value:
        max_value=i
print(max_value)
```

```python
l=[10,2,90,10.5]
min_value=l[0]
for i in l:
    if i<min_value:
        min_value=i
print(min_value)
```

```python
#sum()
l=[10,2,90,10.5]
print(sum(l))
sum=0
for i in l:
    sum=sum+i
print(sum)
    
```

```python
#mean()
l=[1,2,1,2,3,4,4,4,5,6,7,12,1,34,54,34,21,2]
sum=0
count=0
for i in l:
    sum=sum+i
    count=count+1
mean=sum/count
print(round(mean,2))
```

```python
l=["bhopal","madam","decode","naman","agra","nitin"]
l1=[]
for i in l:
    if i==i[::-1]:
        l1.append(i)
print(l1)
   
```

```python
l=["bhopal","madam","decode","naman","agra","nitin"]
l1=[]
for i in l:
    rev=""
    for j in i:
        rev=j+rev
    if i==rev:
        l1.append(i)
print(l1)
```

```python
l=[23,7,12,9]
for i in range(len(l)):
    for j in range(i+1,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l)
```
