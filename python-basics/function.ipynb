```python
def sorted_list(l,reverse=False):
    l1=list(l)
    if reverse==False:
        for i in range(len(l1)):
            for j in range(i+1,len(l1)):
                if l1[i]>l1[j]:
                    l1[i],l1[j]=l1[j],l1[i]
    else:
        for i in range(len(l1)):
            for j in range(i+1,len(l1)):
                if l1[i]<l1[j]:
                    l1[i],l1[j]=l1[j],l1[i]
    return l1
```

```python
lis=[10,2,34,56,13]
print(sorted_list(lis))
```

```python
lis=[10,2,34,56,13]
print(sorted_list(lis,reverse=True))
```

```python
t=(10,2,34,56,13)
print(sorted_list(t))
```

```python
# without parameter without return type
def leap_year():
    year=int(input("enter any number: "))
    if year%4==0:
        if year%100==0:
            if year%400==0:
                print(year,"leap year")
            else:
                print(year,"not a leap year")
        else:
            print(year,"leap year")
    else:
        print(year,"not leap year")
```

```python
leap_year()
```

```python
# with parameter with return type
def leap(year):
    if year%4==0:
        if year%100==0:
            if year%400==0:
                return year,"leap year"
            else:
                return year,"not a leap year"
        else:
            return year,"leap year"
    else:
        return year,"not a leap year"
```

```python
leap(1600)
```

```python
def leap1(year):
    if year%4==0:
        if year%100==0:
            if year%400==0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
```

```python
x=int(input("enter any year: "))
if leap1(x):
    print(x,"leap year")
else:
    print(x,"not a leap year")
```
